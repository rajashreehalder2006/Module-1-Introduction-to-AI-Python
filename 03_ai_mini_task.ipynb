```python
"""
Module 1 - Script 3: Rule-Based Classifier (Foundation of AI Decision Logic)
"""

def rule_based_spam_filter(message):
    spam_keywords = ["free", "win", "cash", "prize", "urgent", "click here", "subscribe"]
    message_clean = message.lower()
    
    detected_keywords = [kw for kw in spam_keywords if kw in message_clean]
    
    # Classification decision
    is_spam = len(detected_keywords) >= 2
    return is_spam, detected_keywords

if __name__ == "__main__":
    emails = [
        "Congratulations! You win a cash prize today. Click here to claim.",
        "Hi, are we still meeting tomorrow at 10 AM for project review?",
        "Urgent: Claim your free gift voucher now!",
    ]

    for email in emails:
        spam_flag, triggers = rule_based_spam_filter(email)
        label = "SPAM" if spam_flag else "HAM (Not Spam)"
        print(f"Text: '{email}'")
        print(f"Prediction: {label} | Triggers: {triggers}\n")
```
