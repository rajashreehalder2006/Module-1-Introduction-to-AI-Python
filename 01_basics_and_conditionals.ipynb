```python
"""
Module 1 - Script 1: Variables, Data Types, and Conditionals
"""

def evaluate_temperature(temp_celsius):
    print(f"Current Temperature: {temp_celsius}°C")
    if temp_celsius > 35:
        return "Hot - Stay hydrated!"
    elif temp_celsius >= 20:
        return "Moderate - Great weather for a walk."
    else:
        return "Cold - Grab a jacket."

if __name__ == "__main__":
    test_temps = [40, 26, 12]
    for temp in test_temps:
        status = evaluate_temperature(temp)
        print(f"Status: {status}\n")
```
