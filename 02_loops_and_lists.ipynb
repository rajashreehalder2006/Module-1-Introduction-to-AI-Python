```python
"""
Module 1 - Script 2: Lists, Iteration, and Aggregations
"""

def analyze_sensor_readings(readings):
    if not readings:
        return None
    
    total = 0
    highest = readings[0]
    lowest = readings[0]

    for val in readings:
        total += val
        if val > highest:
            highest = val
        if val < lowest:
            lowest = val

    average = total / len(readings)
    return {
        "count": len(readings),
        "average": round(average, 2),
        "max": highest,
        "min": lowest
    }

if __name__ == "__main__":
    sensor_data = [23.4, 25.1, 22.8, 26.5, 24.0, 27.2]
    metrics = analyze_sensor_readings(sensor_data)
    
    print("Sensor Data Analysis:")
    for key, value in metrics.items():
        print(f" - {key.capitalize()}: {value}")
```
